```python
REQUIRED_FIELDS = {
    "dataset", "title", "content", "category", "language",
    "source_type", "source", "source_url", "as_of_date"
}

for i, doc in enumerate(ALL_DOCUMENTS):
    missing = REQUIRED_FIELDS - set(doc)
    if missing:
        raise ValueError(f"Document {i} missing: {missing}")
    if doc["dataset"] not in {"general_knowledge", "station", "current_service"}:
        raise ValueError(f"Bad dataset in document {i}")
    if not doc["title"].strip() or not doc["content"].strip():
        raise ValueError(f"Empty document {i}")

required_titles = {
    "Passenger Rights - Timetables, Fares, Notices, Toilets and Insurance",
    "Passenger Complaints and Suggestions to Myanma Railways",
    "Station Amenities May Vary by Station",
    "Special Trains May Operate During Major Holidays",
    "Railway Rolling-Stock Glossary - DEMU, RBE, BEL and BEMU",
    "ORTP e-Receipt Must Be Exchanged Before Departure",
    "Foreign Passengers - Passport Support and Identity Guidance",
}

missing = required_titles - {d["title"] for d in ALL_DOCUMENTS}
if missing:
    raise ValueError(f"Missing required v4.1 knowledge: {missing}")

print("✅ v4.1 validation passed")
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "BAAI/bge-m3",
    trust_remote_code=True,
    device="cpu"
)
print("✅ BGE-M3 ready")
```

Create the same vector-search index on `documents_v4_1`:

- Index name: `vector_index`
- Path: `embedding`
- Dimensions: `1024`
- Similarity: `cosine`

Do not switch the chatbot to this collection until the index is READY and retrieval tests pass.


```python
CONFIRM_V4_1_REBUILD = True

if not CONFIRM_V4_1_REBUILD:
    raise RuntimeError(
        "Review the v4.1 knowledge first, then set CONFIRM_V4_1_REBUILD = True."
    )

mongo_documents = []

for i, doc in enumerate(ALL_DOCUMENTS, start=1):
    print(f"Embedding {i}/{len(ALL_DOCUMENTS)}: {doc['title']}")
    embedding_text = f"{doc['title']}\n\n{doc['content']}"
    embedding = model.encode(
        embedding_text,
        normalize_embeddings=True
    ).tolist()

    mongo_documents.append({
        "dataset": doc["dataset"],
        "title": doc["title"],
        "content": doc["content"],
        "category": doc["category"],
        "language": doc["language"],
        "source_type": doc["source_type"],
        "source": doc["source"],
        "source_url": doc["source_url"],
        "as_of_date": doc["as_of_date"],
        "embedding": embedding,
    })

deleted = collection.delete_many({})
inserted = collection.insert_many(mongo_documents)

print("Deleted old v4.1 docs:", deleted.deleted_count)
print("Inserted:", len(inserted.inserted_ids))
print("Final count:", collection.count_documents({}))
```

```python
def vector_search(query, limit=8, num_candidates=150):
    query_embedding = model.encode(query, normalize_embeddings=True).tolist()

    return list(collection.aggregate([
        {
            "$vectorSearch": {
                "index": "vector_index",
                "path": "embedding",
                "queryVector": query_embedding,
                "numCandidates": num_candidates,
                "limit": limit,
            }
        },
        {
            "$project": {
                "_id": 0,
                "dataset": 1,
                "title": 1,
                "content": 1,
                "category": 1,
                "language": 1,
                "source_type": 1,
                "source": 1,
                "source_url": 1,
                "as_of_date": 1,
                "score": {"$meta": "vectorSearchScore"},
            }
        }
    ]))

TEST_QUESTIONS = [
    "ဘူတာမှာ ပစ္စည်းခဏအပ်ထားလို့ရလား",
    "ရထားလက်မှတ်ရှိရင် ခုံသေချာရလား",
    "ရထားဝန်ဆောင်မှုကို ဘယ်လိုတိုင်ကြားရမလဲ",
    "ပိတ်ရက်မှာ ပြည်ကို အထူးရထားရှိနိုင်လား",
    "DEMU နဲ့ RBE ဘာကွာလဲ",
    "ရထားမတော်တဆမှုဖြစ်ရင် ခရီးသည်အာမခံရှိလား",
    "Do all stations have luggage storage?",
    "Can train stops change over time?",
]

for q in TEST_QUESTIONS:
    print("\n" + "=" * 90)
    print("QUERY:", q)
    for r in vector_search(q, limit=5):
        print(round(r["score"], 4), "|", r["category"], "|", r["title"])
```
